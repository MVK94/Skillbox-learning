## Задача 1. Пунктуация
Напишите программу, которая определяет, сколько различных знаков пунктуации встречается в символьной строке.

К знакам пунктуации относятся символы из набора ".,;:!?". Набор знаков должен храниться в виде множества.

Пример:

Введите строку:

Я! Есть. Грут?! Я, Грут и Есть.

Различные знаки пунктуации: ! . ? ,

Количество различных знаков пунктуации: 4

```python
sym_set = {".",",",";",":","!","?"}
text = "Я! Есть. Грут?! Я, Грут и Есть."

result = {sym for sym in text if not sym.isalnum() and not sym.isspace()}
print("Различные знаки пунктуации: ", result)
print("Количество различных знаков пунктуации: ", len(result))
x = sym_set.difference(result)
print(x)
```

Задача 3. Различные цифры
Напишите программу, которая находит все различные цифры в символьной строке. Для решения используйте множество (цифры будут различные, и поиск во множестве намного быстрее, чем в списке).

Подсказка: можно использовать вот такое сравнение '0' <= x <= '9'



Пример:

Введите строку: ab1n32kz2

Различные цифры строки: 123



```python
import random

nums_1 = [29, 17, 10, 15, 13, 22, 12, 22, 7, 24, 26, 3, 11, 2, 3, 16, 19, 21, 2, 3, 8, 27, 2, 17, 2, 20, 12, 21, 3, 1]
nums_2 = [16, 21, 30, 24, 5, 7, 23, 13, 11, 5, 21, 5, 19, 9, 12, 9, 15, 16, 29, 8, 16, 1, 22, 15, 16, 9, 1, 13, 21, 21]

set_1 = set(nums_1)
set_2 = set(nums_2)

set_1.remove(min(set_1))
set_2.remove(min(set_2))

set_1.add(random.randint(100, 200))
set_2.add(random.randint(100, 200))

union = set_1.union(set_2)
print(union)
intersection = set_1.intersection(set_2)
print(intersection)
difference = set_1.difference(set_2)
print(difference)
```

Задача 3. Различные цифры
Напишите программу, которая находит все различные цифры в символьной строке. Для решения используйте множество (цифры будут различные, и поиск во множестве намного быстрее, чем в списке).

Подсказка: можно использовать вот такое сравнение '0'<=x<='9'



Пример:

Введите строку: ab1n32kz2

Различные цифры строки: 123



```python
text = input("Введите строку: ")

digits = {ch for ch in text if ch.isdigit()}
print(digits)
result = ''.join(sorted(digits))

print("Различные цифры строки:", result)
```
